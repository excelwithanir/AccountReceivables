Remove all the leading and trailing whitespaces from the columns

```python
# Before executing the below code, check the dytype of the columns
for col in df_copy.columns:
    if df_copy[col].dtype == "str":
        df_copy[col] = df_copy[col].astype("string").str.strip()
```

```python
# Standardie CustomerSegment Column with LargeEnterprise, MidMarket and SME
```

```python
df_copy["CustomerSegment"] = (
    df_copy["CustomerSegment"]
      .astype("string")
      .str.strip()
      .str.lower()
      .replace({
          r".*large\s+enterprise.*": "LargeEnterprise",
          r".*mid[-\s]+market.*": "MidMarket",
          r".*sme.*": "SME"
      }, regex=True)
)
```

```python
df_copy["Region"] = df_copy["Region"].str.title()
```
